## Delta Lake MERGE Concept

Delta Lake MERGE is used to perform upsert operations, meaning it can update existing records and insert new records into a dataset.
This concept is widely used in Azure Databricks and modern data engineering pipelines for efficient data updates.

```python
# Create sample new data (simulate incoming data)
new_data = df.sample(5).copy()
```

```python
# Modify values to simulate updates
new_data['Sales'] = new_data['Sales'] + 500
```

```python
# Perform update (like MERGE)
df.update(new_data)
```

```python
# Append new records and remove duplicates (simulate insert + update)
merged_df = pd.concat([df, new_data]).drop_duplicates(subset=['Row ID'], keep='last')
```

```python
# Show result
merged_df.head()
```

### Observation:
The MERGE operation is simulated using Pandas by updating existing records and inserting new ones. This mimics the behavior of Delta Lake MERGE, which is widely used in data engineering workflows.

### SQL Representation of MERGE:

MERGE INTO target_table AS target
USING source_table AS source
ON target.Row_ID = source.Row_ID
WHEN MATCHED THEN
  UPDATE SET target.Sales = source.Sales
WHEN NOT MATCHED THEN
  INSERT *
